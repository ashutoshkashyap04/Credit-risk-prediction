```python
# importing the required libraries and classes

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import (
    RandomForestClassifier,
    AdaBoostClassifier,
    GradientBoostingClassifier,
    StackingClassifier
)
from xgboost import XGBClassifier
```

```python
# load the data
import pandas as pd

train_df = pd.read_csv("../data/processed/train.csv")
test_df = pd.read_csv("../data/processed/test.csv")

print(train_df.shape)
print(test_df.shape)
```

```python
X_train = train_df.drop(columns= "Risk")
y_train = train_df["Risk"]

X_test = test_df.drop(columns= "Risk")
y_test = test_df["Risk"]
```

```python
# scaling the data

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# models dictionary

models = {
    "Logistic Regression" : LogisticRegression(),
    "KNN" : KNeighborsClassifier(),
    "Naive Bayes" : GaussianNB(),
    "Decision Tree" : DecisionTreeClassifier(),
    "Random Forest" : RandomForestClassifier(),
    "SVM" : SVC(probability= True),            # SVC don't return probability by default
    "AdaBoost" : AdaBoostClassifier(),
    "Gradient Boost" : GradientBoostingClassifier(),
    "XGBoost" : XGBClassifier()
}
```

```python
# training models

trained_models = {}

for name, model in models.items():
    model.fit(X_train_scaled, y_train)
    trained_models[name] = model
    
print("All models trained successfully")
```

```python

# Initial model evaluation 

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,          # receiver operating characteristic area under curve
    classification_report,
    confusion_matrix
)
```

```python
result = []

for name, model in trained_models.items():
    
    y_pred = model.predict(X_test_scaled)
    y_prob = model.predict_proba(X_test_scaled)[:,1]
    
    result.append({
        "Model" : name,
        "Accuracy" : accuracy_score(y_test, y_pred),
        "Precision" : precision_score(y_test, y_pred),
        "Recall" : recall_score(y_test, y_pred),
        "F1 Score" : f1_score(y_test, y_pred),
        "ROC-AUC" : roc_auc_score(y_test, y_prob)
    })
    
result_df = pd.DataFrame(result).sort_values(
    by = "F1 Score",
    ascending= False
)

result_df
```

```python
# selecting best 6 models based on f1_score for hyperparameter tuning

models_best_6 = result_df.head(6)
models_best_6
```

```python
# hyperparameter tuning

from sklearn.model_selection import RandomizedSearchCV
```

```python
# function for model tuning
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold

cv = StratifiedKFold(
    n_splits= 5,
    shuffle= True,
    random_state= 42
)

def tune_model(model, params, X_train, y_train):
    rs = RandomizedSearchCV(
        estimator= model,
        param_distributions= params,
        n_iter= 20,
        scoring= 'f1',
        cv= cv,
        verbose= 1,
        n_jobs= -1,  # controls CPU usage, -1 means use all the available CPU cores to make the search faster
        random_state= 42,
        refit = True,      # after finding the best hyperparameters, trains the best model again on the entire training dataset
        return_train_score= False,  # does not include training scores in the search results
        error_score = "raise"       # immediately raises an error if a model fit fails
    )
    
    rs.fit(X_train_scaled, y_train)
    return rs.best_estimator_, rs.best_params_, rs.best_score_
```

```python
import numpy as np

param_distributions = {
    "XGBoost": {
        "n_estimators": [100, 200, 300],
        "max_depth": [3, 5, 7],
        "learning_rate": [0.01, 0.05, 0.1],
    },

    "Random Forest": {
        "n_estimators": [100, 200, 300],
        "max_depth": [None, 5, 10, 15],
        "min_samples_split": [2, 5, 10],
        "min_samples_leaf": [1, 2, 4],
    },

    "Gradient Boost": {
        "n_estimators": [100, 200, 300],
        "max_depth": [2, 3, 5],
        "learning_rate": [0.01, 0.05, 0.1],
    },

    "Decision Tree": {
        "max_depth": [None, 5, 10, 15],
        "min_samples_split": [2, 5, 10],
        "min_samples_leaf": [1, 2, 4],
    },

    "Naive Bayes": {
        "var_smoothing": np.logspace(-12, -7, 10),
    },

    "AdaBoost": {
        "n_estimators": [50, 100, 200],
        "learning_rate": [0.01, 0.1, 0.5, 1.0],
    },
}
```

```python
tuning_results = {}

for name in models_best_6["Model"]:
    print(f"\nTuning: {name}")

    best_model, best_params, best_f1 = tune_model(
        model=models[name],
        params=param_distributions[name],
        X_train=X_train_scaled,
        y_train=y_train
    )

    tuning_results[name] = {
        "Best Model": best_model,
        "Best Parameters": best_params,
        "Best CV F1": best_f1
    }

    print("Best Parameters:", best_params)
    print("Best CV F1:", round(best_f1, 4))
```

```python
tuning_results_df = pd.DataFrame([
    {
        "Model": name,
        "Best CV F1": result["Best CV F1"],
        "Best Parameters": result["Best Parameters"],
        "Best Model": result["Best Model"]
    }
    for name, result in tuning_results.items()
])

tuning_results_df = tuning_results_df.sort_values(
    by="Best CV F1",
    ascending=False
).reset_index(drop=True)


models_best_3 = tuning_results_df.head(3)
```

```python
models_best_3
```

```python
estimators = [
    (row["Model"], row["Best Model"])
    for _, row in models_best_3.iterrows()
]
```

```python
estimators
```

```python
stack_model = StackingClassifier(
    estimators= estimators,
    final_estimator= LogisticRegression(),
    cv = 5,
    n_jobs = -1
)

stack_model.fit(X_train_scaled, y_train)
```

```python

y_pred = stack_model.predict(X_test)
y_prob = stack_model.predict_proba(X_test)[:,1]
```

```python
stacking_f1 = f1_score(y_test, y_pred)
print("Stacking F1", stacking_f1)
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

models_to_evaluate = {
    row["Model"]: row["Best Model"]
    for _, row in models_best_3.iterrows()
}

models_to_evaluate["Stacking"] = stack_model

results = []

for name, model in models_to_evaluate.items():
    y_pred = model.predict(X_test_scaled)
    y_prob = model.predict_proba(X_test_scaled)[:, 1]

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1 Score": f1_score(y_test, y_pred, zero_division=0),
        "ROC-AUC": roc_auc_score(y_test, y_prob)
    })

results_df = pd.DataFrame(results).sort_values(
    by="F1 Score", ascending=False
).reset_index(drop=True)

results_df.round(3)
```
