```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
# data loading
df = pd.read_csv("../data/raw/cleaned_data.csv")
df.head()
```

```python
# One hot encoding of Sex, Housing, Purpose columns

df = pd.get_dummies(
    df,
    columns = ['Sex', 'Housing', 'Purpose'],
    drop_first = True,
    dtype = int
)

df.head()
```

```python
# Savings account mapping

saving_map = {
    "No Account": 0,
    "little": 1,
    "moderate": 2,
    "quite rich": 3,
    "rich": 4
}

df["Saving accounts"] = df["Saving accounts"].map(saving_map)
```

```python
# checking account mapping

checking_map = {
    "No Account" : 0,
    "little" : 1,
    "moderate" : 3,
    "rich" : 3
}

df['Checking account'] = df["Checking account"].map(checking_map)
```

```python
# encoding the target variable (Risk)

df['Risk'] = df['Risk'].map({
    "bad" : 1,
    "good" : 0
})
```

```python
# saving the processed data

df.to_csv("../data/processed/processed_data.csv", index = False)
```
